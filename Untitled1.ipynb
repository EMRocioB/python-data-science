# **3) Analizar los Datos**

```python
#Para calcular el sobreStock

# 1 - Sumar total de exportacion por porducto de la tabla exports_to_paraguay

# Sumar los valores de cada columna de productos
suma_produc_expor = df_exportaciones.drop(columns=['id']).sum()

# Sumar todas las sumas para obtener el total
suma_total_expor = suma_produc_expor.sum()

# Imprimir los totales de cada producto
for producto, total in suma_produc_expor.items():
    print(f"{producto}: {total:.2f}")

print(f"Suma total de exportacion por porducto: {suma_total_expor}")
```

```python
# 2 - Sumar total de ventas por porducto de la tabla sales_in_paraguay

# Sumar los valores de cada columna de productos
suma_produc_venta = df_ventas.drop(columns=['id']).sum()

# Sumar todas las sumas para obtener el total
suma_total_venta = suma_produc_venta.sum()

# Imprimir los totales de cada producto
for producto, total in suma_produc_venta.items():
    print(f"{producto}: {total:.2f}")

print(f"Suma total de ventas por porducto: {suma_total_venta}")
```

```python
# 3 - Calcular del sobreStock: exportacion_totales - ventas_totales

sobrestock = suma_total_expor - suma_total_venta

print(f"Sobrestock: {sobrestock}")
```

```python
# Sumar los valores de cada columna de productos para exportaciones
suma_produc_expor = df_exportaciones.drop(columns=['id']).sum()

# Sumar los valores de cada columna de productos para ventas
suma_produc_venta = df_ventas.drop(columns=['id']).sum()

# Calcular el sobrestock por producto
sobrestock_por_producto = suma_produc_expor - suma_produc_venta

# Imprimir los totales de cada producto
print("Ventas y exportaciones por producto:")
for producto in suma_produc_venta.index:
    venta = suma_produc_venta[producto]
    exportacion = suma_produc_expor[producto]
    sobrestock = sobrestock_por_producto[producto]
    print(f"{producto}: Ventas: {venta:.2f}, Exportaciones: {exportacion:.2f}, Sobrestock: {sobrestock:.2f}")

# Calcular e imprimir el sobrestock total
sobrestock_total = sobrestock_por_producto.sum()

print(f"Sobrestock total: {sobrestock_total:.2f}")

```

```python
# Identificación de Productos con Sobrestock

sobrestock_por_producto = suma_produc_expor - suma_produc_venta
print(sobrestock_por_producto.sort_values(ascending=False))
```

```python
porcentaje_sobrestock = (sobrestock_por_producto / suma_produc_expor) * 100
print(porcentaje_sobrestock.sort_values(ascending=False))
```

# **4) Visualizar Datos**

```python
# Selección de los 10 productos con mayor sobrestock
top_10_sobrestock = sobrestock_por_producto.nlargest(10)

# Crear una figura con dos subgráficos (uno al lado del otro)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 6))

# Gráfico de barras en el primer subgráfico
top_10_sobrestock.plot(kind='bar', ax=ax1, color='skyblue', edgecolor='black')
ax1.set_title('Top 10 Productos con Mayor Sobrestock (Barra)')
ax1.set_ylabel('Sobrestock (en pesos)')
ax1.set_xticklabels(top_10_sobrestock.index, rotation=45)

# Gráfico de líneas en el segundo subgráfico
ax2.plot(top_10_sobrestock.index, top_10_sobrestock.values, marker='o', linestyle='-', color='b')
ax2.set_title('Top 10 Productos con Mayor Sobrestock (Línea)')
ax2.set_xlabel('Producto')
ax2.set_ylabel('Sobrestock (en pesos)')
ax2.grid(True)

# Ajustar el diseño para que no se corten las etiquetas
plt.tight_layout()

# Mostrar los gráficos
plt.show()
```
