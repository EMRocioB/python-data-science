```python
import PyPDF2
import os

def merge_pdfs(pdf_files, output_file):
    merger = PyPDF2.PdfMerger()

    for pdf_file in pdf_files:
        with open(pdf_file, 'rb') as file:
            merger.append(file)

    with open(output_file, 'wb') as output:
        merger.write(output)

pdf_files_to_merge = ['E:\\unirpdf\\1.pdf', 'E:\\unirpdf\\2.pdf', 'E:\\unirpdf\\3.pdf', 'E:\\unirpdf\\4.pdf']
output_pdf_file = 'E:\\unirpdf\\5.pdf'

merge_pdfs(pdf_files_to_merge, output_pdf_file)
```

```python
from google.colab import files
import PyPDF2

def merge_pdfs(pdf_files, output_file):
    merger = PyPDF2.PdfMerger()

    for pdf_file in pdf_files:
        merger.append(pdf_file)

    with open(output_file, 'wb') as output:
        merger.write(output)

# Mostrar el widget para cargar varios archivos
uploaded = files.upload()

# Obtener los nombres de los archivos cargados
pdf_files_to_merge = list(uploaded.keys())

# Especificar el nombre del archivo de salida
output_pdf_file = 'merged_file.pdf'

# Fusionar archivos PDF
merge_pdfs(pdf_files_to_merge, output_pdf_file)

```
